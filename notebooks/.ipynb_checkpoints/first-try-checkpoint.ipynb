```python
import pandas as pd
import os

# Get the directory where this notebook is located
notebook_dir = os.getcwd()

# Join paths to safely find the file
# We go "up" one level (..) then into "data", then the file
file_path = os.path.join(notebook_dir, "..", "data", "filteredvehpub.csv")

# Print it to verify it looks right
print(f"Trying to load: {os.path.abspath(file_path)}")

# Load the file
df = pd.read_csv(file_path)
df.head()
```

```python
df.columns
```

```python
import numpy as np
invalid_values = [-9, -8, -7, -88, 99, "99", "XX", "xx", "XX ", "-88", "-9", "-8", "-7"]
df = df.replace(invalid_values, np.nan)
df = df.dropna()
```

```python
df["VEHTYPE"] = df["VEHTYPE"].replace([5, 6], np.nan)
df = df.dropna()
```

```python
counts = df["MAKE"].value_counts()
counts.describe()
```

```python
df["MAKE"] = df["MAKE"].astype(str)   # convert everything to string first
df["MAKE"] = df["MAKE"].str.strip()   # remove whitespace
df["MAKE"] = df["MAKE"].astype(int)   # convert to integer

counts = df["MAKE"].value_counts()
rare_makes = counts[counts < 1000].index

df["MAKE"] = df["MAKE"].where(~df["MAKE"].isin(rare_makes), 98)
df["MAKE"].value_counts()
```

```python
#df.to_csv("processed_data.csv", index=False)
```

```python
df.columns.tolist()
```

```python
feature_cols = [
    "HHFAMINC",
    "HHSIZE",
    "LIF_CYC",
    "HHSTATE",
    "URBAN"
]
target_col = "MAKE"
X = df[feature_cols]
y = df[target_col]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

cat_cols = ["HHFAMINC", "LIF_CYC", "HHSTATE", "URBAN"]
num_cols = ["HHSIZE"]   # only numeric column

preprocess = ColumnTransformer(
    transformers=[
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), cat_cols),
        ("num", "passthrough", num_cols)
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1,           # use all CPU cores
    min_samples_split=4,
    min_samples_leaf=2
)

pipe = Pipeline([
    ("preprocess", preprocess),   # your ColumnTransformer from Step 7
    ("rf", rf)
])

pipe
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

cv_acc = cross_val_score(
    pipe,
    X, y,
    cv=cv,
    scoring='accuracy',
    n_jobs=-1
)

print("Cross-validated Top-1 Accuracy:", cv_acc.mean(), "+/-", cv_acc.std())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
pipe.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score

y_pred = pipe.predict(X_test)

test_acc = accuracy_score(y_test, y_pred)
print("Test Top-1 Accuracy:", test_acc)
```

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import numpy as np

cm = confusion_matrix(y_test, y_pred)

labels = sorted(y.unique())

fig, ax = plt.subplots(figsize=(10, 10))
im = ax.imshow(cm, cmap="Blues")
plt.colorbar(im)

ax.set_xticks(np.arange(len(labels)))
ax.set_yticks(np.arange(len(labels)))
ax.set_xticklabels(labels, rotation=45, ha="right")
ax.set_yticklabels(labels)

ax.set_xlabel("Predicted")
ax.set_ylabel("True")
ax.set_title("Confusion Matrix (Top-1 Predictions)")

plt.tight_layout()
plt.show()
```
